```python
# Identify numerical columns with missing values
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

print("Filling missing numerical values with their column means:")
for col in numerical_cols:
    if df[col].isnull().any():
        mean_val = df[col].mean()
        df[col].fillna(mean_val, inplace=True)
        print(f"  - Missing values in column '{col}' filled with mean: {mean_val:.2f}")

# Display the first few rows of the DataFrame after imputation
print("\nDataFrame after filling numerical missing values with mean:")
display(df.head())
```
